```python
import pandas as pd
import numpy as np
import joblib
import xgboost as xgb

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import StratifiedGroupKFold, cross_val_score
from sklearn.metrics import (
    classification_report,
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix
)


# ============================================================
# 1. SETTINGS
# ============================================================

DATA_PATH = r"C:\Users\joshh\Machine_Learning\Audiology_Project\Audio_Data.csv"
MODEL_PATH = "audiology_model.joblib"

FREQUENCIES = [
    250, 500, 750, 1000, 1500,
    2000, 3000, 4000, 6000, 8000
]

LEFT_AIR = [f"L{f}C" for f in FREQUENCIES]
RIGHT_AIR = [f"R{f}C" for f in FREQUENCIES]

LEFT_BONE = [f"BL{f}C" for f in FREQUENCIES]
RIGHT_BONE = [f"BR{f}C" for f in FREQUENCIES]


# ============================================================
# 2. LOAD DATA
# ============================================================

df = pd.read_csv(DATA_PATH)

df["INVESTDATE"] = pd.to_datetime(
    df["INVESTDATE"],
    format="%d/%m/%Y",
    errors="coerce"
)


# ============================================================
# 3. REPLACE 0 WITH MISSING VALUES
# ============================================================

air_columns = LEFT_AIR + RIGHT_AIR

df[air_columns] = df[air_columns].replace(0, np.nan)


# ============================================================
# 4. REMOVE DUPLICATE PATIENT/DATE/CURVE RECORDS
# ============================================================

all_audiogram_columns = air_columns

df["_missing_count"] = df[all_audiogram_columns].isna().sum(axis=1)

df = (
    df.sort_values(
        ["Unique PID", "INVESTDATE", "CURVE", "_missing_count"]
    )
    .drop_duplicates(
        subset=["Unique PID", "INVESTDATE", "CURVE"],
        keep="first"
    )
    .drop(columns="_missing_count")
)


# ============================================================
# 5. SEPARATE AIR AND BONE RECORDS
# ============================================================

air_df = df[df["CURVE"] == 0].copy()
bone_df = df[df["CURVE"] == 10].copy()


# ============================================================
# 6. RENAME BONE CONDUCTION COLUMNS
# ============================================================

bone_rename = {}

for f in FREQUENCIES:
    bone_rename[f"L{f}C"] = f"BL{f}C"
    bone_rename[f"R{f}C"] = f"BR{f}C"

bone_df = bone_df.rename(columns=bone_rename)


# ============================================================
# 7. MERGE AIR + BONE RECORDS
# ============================================================

bone_features = (
    ["Unique PID", "INVESTDATE", "CURVE"]
    + LEFT_BONE
    + RIGHT_BONE
)

bone_df = bone_df[bone_features]

df_model = air_df.merge(
    bone_df,
    on=["Unique PID", "INVESTDATE"],
    how="inner"
)


# ============================================================
# 8. CALCULATE MEANS
# ============================================================

df_model["Right_Mean"] = df_model[RIGHT_AIR].mean(axis=1)
df_model["Left_Mean"] = df_model[LEFT_AIR].mean(axis=1)

df_model["BRight_Mean"] = df_model[RIGHT_BONE].mean(axis=1)
df_model["BLeft_Mean"] = df_model[LEFT_BONE].mean(axis=1)


# ============================================================
# 9. AIR/BONE CONDUCTION DIFFERENCE
# ============================================================

df_model["RAC_BC"] = abs(
    df_model["Right_Mean"] -
    df_model["BRight_Mean"]
)

df_model["LAC_BC"] = abs(
    df_model["Left_Mean"] -
    df_model["BLeft_Mean"]
)


# ============================================================
# 10. EAR ASYMMETRY
# ============================================================

df_model["Ear_Asymmetry"] = abs(
    df_model["Right_Mean"] -
    df_model["Left_Mean"]
)


# ============================================================
# 11. LOW/HIGH FREQUENCY MEANS
# ============================================================

LOW_FREQUENCIES = FREQUENCIES[:5]
HIGH_FREQUENCIES = FREQUENCIES[5:]

LEFT_AIR_LOW = [f"L{f}C" for f in LOW_FREQUENCIES]
RIGHT_AIR_LOW = [f"R{f}C" for f in LOW_FREQUENCIES]

LEFT_AIR_HIGH = [f"L{f}C" for f in HIGH_FREQUENCIES]
RIGHT_AIR_HIGH = [f"R{f}C" for f in HIGH_FREQUENCIES]

LEFT_BONE_LOW = [f"BL{f}C" for f in LOW_FREQUENCIES]
RIGHT_BONE_LOW = [f"BR{f}C" for f in LOW_FREQUENCIES]

LEFT_BONE_HIGH = [f"BL{f}C" for f in HIGH_FREQUENCIES]
RIGHT_BONE_HIGH = [f"BR{f}C" for f in HIGH_FREQUENCIES]


df_model["LBLeft_Mean"] = df_model[LEFT_AIR_LOW].mean(axis=1)
df_model["LBRight_Mean"] = df_model[RIGHT_AIR_LOW].mean(axis=1)

df_model["HBLeft_Mean"] = df_model[LEFT_AIR_HIGH].mean(axis=1)
df_model["HBRight_Mean"] = df_model[RIGHT_AIR_HIGH].mean(axis=1)

df_model["LLeft_Mean"] = df_model[LEFT_BONE_LOW].mean(axis=1)
df_model["LRight_Mean"] = df_model[RIGHT_BONE_LOW].mean(axis=1)

df_model["HLeft_Mean"] = df_model[LEFT_BONE_HIGH].mean(axis=1)
df_model["HRight_Mean"] = df_model[RIGHT_BONE_HIGH].mean(axis=1)


# ============================================================
# 12. REMOVE UNUSABLE TARGET ROWS
# ============================================================

df_model = df_model.dropna(
    subset=["Treatment Cat"]
).copy()


# ============================================================
# 13. YOUR ORIGINAL EXCLUSION RULE
# ============================================================

df_model = df_model[
    ~(
        (df_model["Treatment Cat"] == "No Treatment")
        &
        (
            (df_model["Right_Mean"] > 30)
            |
            (df_model["Left_Mean"] > 30)
        )
    )
].copy()


# ============================================================
# 14. DEFINE FEATURES EXPLICITLY
# ============================================================

engineered_features = [
    "Right_Mean",
    "Left_Mean",
    "RAC_BC",
    "LAC_BC",
    "Ear_Asymmetry",
    "BLeft_Mean",
    "BRight_Mean",
    "HBRight_Mean",
    "HBLeft_Mean",
    "HRight_Mean",
    "HLeft_Mean",
    "LBRight_Mean",
    "LBLeft_Mean",
    "LRight_Mean",
    "LLeft_Mean"
]

raw_features = (
    LEFT_AIR
    + RIGHT_AIR
    + LEFT_BONE
    + RIGHT_BONE
)

feature_cols = raw_features + engineered_features


X = df_model[feature_cols].copy()
y = df_model["Treatment Cat"].copy()

groups = df_model["Unique PID"]


# ============================================================
# 15. ENCODE TARGET
# ============================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("\nTarget classes:")
for number, name in enumerate(label_encoder.classes_):
    print(f"{number}: {name}")


# ============================================================
# 16. XGBOOST MODEL
# ============================================================

model = xgb.XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    min_child_weight=3,
    reg_alpha=0.1,
    reg_lambda=1.0,
    objective="multi:softprob",
    eval_metric="mlogloss",
    random_state=1415,
    n_jobs=-1
)


# ============================================================
# 17. PATIENT-LEVEL CROSS VALIDATION
# ============================================================

cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=1415
)

print("\nRunning patient-level cross-validation...")

cv_scores = cross_val_score(
    model,
    X,
    y_encoded,
    cv=cv,
    groups=groups,
    scoring="balanced_accuracy",
    n_jobs=-1
)

print("\nCross-validation balanced accuracy:")
print(cv_scores)

print(
    f"\nMean balanced accuracy: "
    f"{cv_scores.mean():.3f}"
)


# ============================================================
# 18. TRAIN FINAL MODEL ON ALL DATA
# ============================================================

model.fit(X, y_encoded)


# ============================================================
# 19. SAVE EVERYTHING REQUIRED BY STREAMLIT
# ============================================================

model_package = {
    "model": model,
    "label_encoder": label_encoder,
    "feature_cols": feature_cols,
    "frequencies": FREQUENCIES,
    "raw_features": raw_features,
    "engineered_features": engineered_features
}

joblib.dump(
    model_package,
    MODEL_PATH
)

print(f"\nModel saved to: {MODEL_PATH}")
print(f"Number of observations: {len(df_model)}")
print(f"Number of features: {len(feature_cols)}")
```

```python
import streamlit as st
import pandas as pd
import numpy as np
import joblib


# ============================================================
# PAGE CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="Audiology Treatment Predictor",
    page_icon="👂",
    layout="wide"
)


# ============================================================
# LOAD MODEL
# ============================================================

@st.cache_resource
def load_model():
    return joblib.load("audiology_model.joblib")


model_package = load_model()

model = model_package["model"]
label_encoder = model_package["label_encoder"]
feature_cols = model_package["feature_cols"]


# ============================================================
# FREQUENCIES
# ============================================================

FREQUENCIES = [
    250, 500, 750, 1000, 1500,
    2000, 3000, 4000, 6000, 8000
]


# ============================================================
# HELPER FUNCTION
# ============================================================

def get_input_value(label, key):
    """
    Number input allowing missing values.
    """

    return st.number_input(
        label,
        min_value=-20.0,
        max_value=120.0,
        value=0.0,
        step=5.0,
        key=key,
        help="Enter threshold in dB HL. Use 0 if no threshold is available."
    )


# ============================================================
# FEATURE ENGINEERING
# ============================================================

def create_features(values):
    """
    Convert user-entered audiogram values into
    exactly the same features used during training.
    """

    row = {}

    # --------------------------------------------------------
    # AIR CONDUCTION
    # --------------------------------------------------------

    for frequency in FREQUENCIES:

        row[f"L{frequency}C"] = values[
            f"L{frequency}C"
        ]

        row[f"R{frequency}C"] = values[
            f"R{frequency}C"
        ]

    # --------------------------------------------------------
    # BONE CONDUCTION
    # --------------------------------------------------------

    for frequency in FREQUENCIES:

        row[f"BL{frequency}C"] = values[
            f"BL{frequency}C"
        ]

        row[f"BR{frequency}C"] = values[
            f"BR{frequency}C"
        ]

    # --------------------------------------------------------
    # CONVERT 0 TO NAN
    # --------------------------------------------------------

    for column in row:

        if row[column] == 0:
            row[column] = np.nan

    # --------------------------------------------------------
    # DATAFRAME
    # --------------------------------------------------------

    df = pd.DataFrame([row])

    # --------------------------------------------------------
    # MEANS
    # --------------------------------------------------------

    left_air = [
        f"L{f}C"
        for f in FREQUENCIES
    ]

    right_air = [
        f"R{f}C"
        for f in FREQUENCIES
    ]

    left_bone = [
        f"BL{f}C"
        for f in FREQUENCIES
    ]

    right_bone = [
        f"BR{f}C"
        for f in FREQUENCIES
    ]

    df["Right_Mean"] = df[right_air].mean(axis=1)
    df["Left_Mean"] = df[left_air].mean(axis=1)

    df["BRight_Mean"] = df[right_bone].mean(axis=1)
    df["BLeft_Mean"] = df[left_bone].mean(axis=1)

    # --------------------------------------------------------
    # AIR/BONE DIFFERENCE
    # --------------------------------------------------------

    df["RAC_BC"] = abs(
        df["Right_Mean"] -
        df["BRight_Mean"]
    )

    df["LAC_BC"] = abs(
        df["Left_Mean"] -
        df["BLeft_Mean"]
    )

    # --------------------------------------------------------
    # EAR ASYMMETRY
    # --------------------------------------------------------

    df["Ear_Asymmetry"] = abs(
        df["Right_Mean"] -
        df["Left_Mean"]
    )

    # --------------------------------------------------------
    # FREQUENCY GROUPS
    # --------------------------------------------------------

    low = FREQUENCIES[:5]
    high = FREQUENCIES[5:]

    left_air_low = [
        f"L{f}C" for f in low
    ]

    right_air_low = [
        f"R{f}C" for f in low
    ]

    left_air_high = [
        f"L{f}C" for f in high
    ]

    right_air_high = [
        f"R{f}C" for f in high
    ]

    left_bone_low = [
        f"BL{f}C" for f in low
    ]

    right_bone_low = [
        f"BR{f}C" for f in low
    ]

    left_bone_high = [
        f"BL{f}C" for f in high
    ]

    right_bone_high = [
        f"BR{f}C" for f in high
    ]

    # --------------------------------------------------------
    # LOW FREQUENCY AIR
    # --------------------------------------------------------

    df["LBLeft_Mean"] = df[
        left_air_low
    ].mean(axis=1)

    df["LBRight_Mean"] = df[
        right_air_low
    ].mean(axis=1)

    # --------------------------------------------------------
    # HIGH FREQUENCY AIR
    # --------------------------------------------------------

    df["HBLeft_Mean"] = df[
        left_air_high
    ].mean(axis=1)

    df["HBRight_Mean"] = df[
        right_air_high
    ].mean(axis=1)

    # --------------------------------------------------------
    # LOW FREQUENCY BONE
    # --------------------------------------------------------

    df["LLeft_Mean"] = df[
        left_bone_low
    ].mean(axis=1)

    df["LRight_Mean"] = df[
        right_bone_low
    ].mean(axis=1)

    # --------------------------------------------------------
    # HIGH FREQUENCY BONE
    # --------------------------------------------------------

    df["HLeft_Mean"] = df[
        left_bone_high
    ].mean(axis=1)

    df["HRight_Mean"] = df[
        right_bone_high
    ].mean(axis=1)

    # --------------------------------------------------------
    # ENSURE EXACT FEATURE ORDER
    # --------------------------------------------------------

    df = df[feature_cols]

    return df


# ============================================================
# HEADER
# ============================================================

st.title("👂 Audiology Treatment Predictor")

st.markdown(
    """
    Enter the patient's air-conduction and bone-conduction
    thresholds below. The trained XGBoost model will calculate
    the derived audiological features and return its predicted
    treatment category.
    """
)

st.info(
    "This application is a machine-learning decision-support "
    "tool and should not be used as a substitute for clinical "
    "assessment or professional judgement."
)


# ============================================================
# PATIENT INPUT
# ============================================================

with st.form("audiogram_form"):

    st.subheader("Air Conduction")

    st.write(
        "Enter thresholds in dB HL. "
        "Use 0 where your original dataset uses 0 to represent missing data."
    )

    col1, col2 = st.columns(2)

    values = {}

    # --------------------------------------------------------
    # LEFT AIR
    # --------------------------------------------------------

    with col1:

        st.markdown("### Left Ear")

        for frequency in FREQUENCIES:

            values[f"L{frequency}C"] = get_input_value(
                f"Left {frequency} Hz",
                f"L{frequency}C"
            )

    # --------------------------------------------------------
    # RIGHT AIR
    # --------------------------------------------------------

    with col2:

        st.markdown("### Right Ear")

        for frequency in FREQUENCIES:

            values[f"R{frequency}C"] = get_input_value(
                f"Right {frequency} Hz",
                f"R{frequency}C"
            )


    st.divider()

    st.subheader("Bone Conduction")

    col1, col2 = st.columns(2)

    # --------------------------------------------------------
    # LEFT BONE
    # --------------------------------------------------------

    with col1:

        st.markdown("### Left Ear")

        for frequency in FREQUENCIES:

            values[f"BL{frequency}C"] = get_input_value(
                f"Left Bone {frequency} Hz",
                f"BL{frequency}C"
            )

    # --------------------------------------------------------
    # RIGHT BONE
    # --------------------------------------------------------

    with col2:

        st.markdown("### Right Ear")

        for frequency in FREQUENCIES:

            values[f"BR{frequency}C"] = get_input_value(
                f"Right Bone {frequency} Hz",
                f"BR{frequency}C"
            )


    st.divider()

    submitted = st.form_submit_button(
        "Predict Treatment Category",
        type="primary",
        use_container_width=True
    )


# ============================================================
# PREDICTION
# ============================================================

if submitted:

    try:

        X_new = create_features(values)

        prediction_encoded = model.predict(X_new)[0]

        prediction = label_encoder.inverse_transform(
            [prediction_encoded]
        )[0]

        # ----------------------------------------------------
        # PROBABILITIES
        # ----------------------------------------------------

        probabilities = model.predict_proba(X_new)[0]

        probability_df = pd.DataFrame({
            "Treatment Category": label_encoder.classes_,
            "Probability": probabilities
        })

        probability_df = probability_df.sort_values(
            "Probability",
            ascending=False
        )

        # ----------------------------------------------------
        # MAIN RESULT
        # ----------------------------------------------------

        st.success(
            f"Predicted Treatment Category: **{prediction}**"
        )

        st.subheader("Prediction probabilities")

        probability_display = probability_df.copy()

        probability_display["Probability"] = (
            probability_display["Probability"] * 100
        ).round(1)

        probability_display = probability_display.rename(
            columns={
                "Probability": "Probability (%)"
            }
        )

        st.dataframe(
            probability_display,
            use_container_width=True,
            hide_index=True
        )

        # ----------------------------------------------------
        # PROBABILITY CHART
        # ----------------------------------------------------

        chart_df = probability_df.set_index(
            "Treatment Category"
        )

        st.bar_chart(
            chart_df["Probability"]
        )

        # ----------------------------------------------------
        # CALCULATED FEATURES
        # ----------------------------------------------------

        st.subheader("Calculated audiological features")

        calculated_features = X_new[
            [
                "Right_Mean",
                "Left_Mean",
                "BRight_Mean",
                "BLeft_Mean",
                "RAC_BC",
                "LAC_BC",
                "Ear_Asymmetry",
                "HBRight_Mean",
                "HBLeft_Mean",
                "HRight_Mean",
                "HLeft_Mean",
                "LBRight_Mean",
                "LBLeft_Mean",
                "LRight_Mean",
                "LLeft_Mean"
            ]
        ].T

        calculated_features.columns = [
            "Value"
        ]

        st.dataframe(
            calculated_features,
            use_container_width=True
        )

        # ----------------------------------------------------
        # INPUT DATA
        # ----------------------------------------------------

        with st.expander("View model input data"):

            st.dataframe(
                X_new.T,
                use_container_width=True
            )

    except Exception as e:

        st.error(
            f"An error occurred while making the prediction: {e}"
        )
```
