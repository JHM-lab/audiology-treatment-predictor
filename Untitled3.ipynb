```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.metrics import classification_report, accuracy_score, roc_auc_score, roc_curve, auc
from sklearn.utils.class_weight import compute_sample_weight

# LightGBM Classifier
from lightgbm import LGBMClassifier

# 1. Load Data & Parse Dates
df = pd.read_csv(r"C:\Users\joshh\Machine_Learning\Audiology_Project\Audio_Data.csv")
df['INVESTDATE'] = pd.to_datetime(df['INVESTDATE'], format="%d/%m/%Y")

# Define frequency columns
LeftSC = [f'L{f}C' for f in [250, 500, 750, 1000, 1500, 2000, 3000, 4000, 6000, 8000]]
RightSC = [f'R{f}C' for f in [250, 500, 750, 1000, 1500, 2000, 3000, 4000, 6000, 8000]]

# Replace 0s with NaN for audiological thresholds
df[LeftSC] = df[LeftSC].replace(0, np.nan)
df[RightSC] = df[RightSC].replace(0, np.nan)

# Calculate overall means
df['Right_Mean'] = df[RightSC].mean(axis=1)
df['Left_Mean'] = df[LeftSC].mean(axis=1)

# Sort and keep the record with the fewest NaNs per visit
df = df.sort_values(by=['Unique PID', 'INVESTDATE', 'CURVE'])
df['_nan_count'] = df.isnull().sum(axis=1)
df = df.loc[df.groupby(['Unique PID', 'INVESTDATE', 'CURVE'])['_nan_count'].idxmin()].drop(columns=['_nan_count'])

# 2. Grouped Shifting (Prevents cross-patient data leakage)
shift_cols = LeftSC + RightSC + ['INVESTDATE', 'CURVE', 'Right_Mean', 'Left_Mean']
shifted_df = df[shift_cols].groupby(df['Unique PID']).shift(-1)

# Rename shifted columns
rename_dict = {col: f'B{col}' for col in LeftSC + RightSC}
rename_dict.update({
    'INVESTDATE': 'next_DATE', 
    'CURVE': 'next_CURVE', 
    'Right_Mean': 'next_Right_Mean', 
    'Left_Mean': 'next_Left_Mean'
})
shifted_df = shifted_df.rename(columns=rename_dict)

# Join shifted columns back to main dataframe safely
df = pd.concat([df, shifted_df], axis=1)

# Assign next_PID since patients stay the same within their group
df['next_PID'] = df['Unique PID']

# Define Baseline frequency columns
BLeftSC = [f'BL{f}C' for f in [250, 500, 750, 1000, 1500, 2000, 3000, 4000, 6000, 8000]]
BRightSC = [f'BR{f}C' for f in [250, 500, 750, 1000, 1500, 2000, 3000, 4000, 6000, 8000]]

# 3. Condition & BC Calculations
condition = (
    (df['Unique PID'] == df['next_PID']) & 
    (df['INVESTDATE'] == df['next_DATE']) & 
    (df['CURVE'] == 0) & 
    (df['next_CURVE'] == 10)
)

df['RAC_BC'] = np.where(condition, np.abs(df['Right_Mean'] - df['next_Right_Mean']), np.nan)
df['LAC_BC'] = np.where(condition, np.abs(df['Left_Mean'] - df['next_Left_Mean']), np.nan)

df_end = df.dropna(subset=['RAC_BC', 'LAC_BC'], how='all').copy().reset_index(drop=True)
df_end['Ear_Asymmetry'] = np.abs(df_end['Right_Mean'] - df_end['Left_Mean'])

df_end.drop(columns=['next_PID', 'next_DATE', 'next_CURVE', 'next_Right_Mean', 'next_Left_Mean'], inplace=True, errors='ignore')

# 4. Frequency Subsets (Low vs High)
LLeftSC, LRightSC = LeftSC[:5], RightSC[:5]
HLeftSC, HRightSC = LeftSC[5:], RightSC[5:]
LBLeftSC, LBRightSC = BLeftSC[:5], BRightSC[:5]
HBLeftSC, HBRightSC = BLeftSC[5:], BRightSC[5:]

# Baseline (B) and Follow-up Means
df_end['BRight_Mean'] = df_end[BRightSC].mean(axis=1)
df_end['BLeft_Mean'] = df_end[BLeftSC].mean(axis=1)

df_end['HBRight_Mean'] = df_end[HBRightSC].mean(axis=1)
df_end['HBLeft_Mean'] = df_end[HBLeftSC].mean(axis=1)
df_end['HRight_Mean'] = df_end[HRightSC].mean(axis=1) 
df_end['HLeft_Mean'] = df_end[HLeftSC].mean(axis=1)    

df_end['LBRight_Mean'] = df_end[LBRightSC].mean(axis=1)
df_end['LBLeft_Mean'] = df_end[LBLeftSC].mean(axis=1)
df_end['LRight_Mean'] = df_end[LRightSC].mean(axis=1) 
df_end['LLeft_Mean'] = df_end[LLeftSC].mean(axis=1)    

# 5. Filtering Test Data
df_test = df_end[~((df_end['Treatment Cat'] == 'No Treatment') & ((df_end['Right_Mean'] > 40) | (df_end['Left_Mean'] > 40)))].copy()

# 6. Feature Engineering & Selection
engineered_cols = [
    'Right_Mean', 'Left_Mean', 'RAC_BC', 'LAC_BC', 'Ear_Asymmetry', 
    'BLeft_Mean', 'BRight_Mean', 'HBRight_Mean', 'HBLeft_Mean', 
    'HRight_Mean', 'HLeft_Mean', 'LBRight_Mean', 'LBLeft_Mean', 
    'LRight_Mean', 'LLeft_Mean'
]
frequency_cols = [c for c in df_test.columns if (c.startswith('R') or c.startswith('L') or c.startswith('B')) and c not in engineered_cols]
feature_cols = frequency_cols + engineered_cols

# Define X and y
X = df_test[feature_cols]
y = df_test['Treatment Cat']

# 7. Label Encoding & Train/Test Split
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
target_names = label_encoder.classes_

X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=1415, stratify=y_encoded
)

# Binarize labels for multi-class ROC calculation
y_test_binarized = label_binarize(y_test, classes=np.unique(y_encoded))

# 8. Train LightGBM (Tuned for Class Imbalance)
print("=" * 60)
print("Training Model: LightGBM")
print("=" * 60)

model = LGBMClassifier(
    n_estimators=150, 
    max_depth=5, 
    learning_rate=0.1, 
    class_weight='balanced',             
    random_state=1415, 
    verbose=-1
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)

# 9. Evaluate Metrics & Print Results
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
roc_auc_macro = roc_auc_score(y_test_binarized, y_prob, average='macro', multi_class='ovr')
print(f"Macro ROC-AUC (OvR): {roc_auc_macro:.4f}\n")

print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=target_names))

# 10. Plot ROC Curves
plt.figure(figsize=(8, 6))
for i, class_name in enumerate(target_names):
    fpr, tpr, _ = roc_curve(y_test_binarized[:, i], y_prob[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, lw=2, label=f'{class_name} (AUC = {roc_auc:.2f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('LightGBM - Multi-Class ROC Curves')
plt.legend(loc="lower right", fontsize=9)
plt.tight_layout()
plt.show()

# ==========================================
# STEP 1: Save Artifacts for Streamlit App
# ==========================================
joblib.dump(model, 'lightgbm_audiology_model.pkl')
joblib.dump(label_encoder, 'label_encoder.pkl')
joblib.dump(feature_cols, 'feature_cols.pkl')
df_test.to_csv('historical_reference_data.csv', index=False)

print("\n[SUCCESS] Model, label encoder, feature columns, and historical reference data saved successfully for Streamlit integration!")
```

```python
import streamlit as st
import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

# Page configuration
st.set_page_config(page_title="Audiology Treatment Predictor & Comparator", layout="wide")

@st.cache_resource
def load_artifacts():
    model = joblib.load('lightgbm_audiology_model.pkl')
    le = joblib.load('label_encoder.pkl')
    features = joblib.load('feature_cols.pkl')
    ref_df = pd.read_csv('historical_reference_data.csv')
    return model, le, features, ref_df

model, label_encoder, feature_cols, ref_df = load_artifacts()

st.title("🎧 Audiology Treatment Predictor & Dataset Comparator")
st.markdown("Enter patient audiological thresholds below to predict the recommended treatment category and compare the patient's profile against historical patient distributions.")

# Sidebar for User Inputs
st.sidebar.header("Patient Audiological Data Entry")

frequencies = [250, 500, 750, 1000, 1500, 2000, 3000, 4000, 6000, 8000]

st.sidebar.subheader("Left Ear Thresholds (dB)")
left_inputs = {}
for f in frequencies:
    left_inputs[f'L{f}C'] = st.sidebar.number_input(f"Left {f} Hz", min_value=0.0, max_value=120.0, value=25.0, step=5.0)

st.sidebar.subheader("Right Ear Thresholds (dB)")
right_inputs = {}
for f in frequencies:
    right_inputs[f'R{f}C'] = st.sidebar.number_input(f"Right {f} Hz", min_value=0.0, max_value=120.0, value=25.0, step=5.0)

st.sidebar.subheader("Baseline / Prior Measures (dB)")
b_left_inputs = {}
for f in frequencies:
    b_left_inputs[f'BL{f}C'] = st.sidebar.number_input(f"Baseline Left {f} Hz", min_value=0.0, max_value=120.0, value=25.0, step=5.0)

b_right_inputs = {}
for f in frequencies:
    b_right_inputs[f'BR{f}C'] = st.sidebar.number_input(f"Baseline Right {f} Hz", min_value=0.0, max_value=120.0, value=25.0, step=5.0)

# Construct user feature dictionary
user_data = {**left_inputs, **right_inputs, **b_left_inputs, **b_right_inputs}
user_df = pd.DataFrame([user_data])

# Calculate engineered features mirroring your training script
user_df['Right_Mean'] = user_df[[f'R{f}C' for f in frequencies]].mean(axis=1)
user_df['Left_Mean'] = user_df[[f'L{f}C' for f in frequencies]].mean(axis=1)
user_df['BRight_Mean'] = user_df[[f'BR{f}C' for f in frequencies]].mean(axis=1)
user_df['BLeft_Mean'] = user_df[[f'BL{f}C' for f in frequencies]].mean(axis=1)

# Frequency bands splits
H_freqs = [3000, 4000, 6000, 8000]
L_freqs = [250, 500, 750, 1000, 1500, 2000]
user_df['HRight_Mean'] = user_df[[f'R{f}C' for f in H_freqs]].mean(axis=1)
user_df['HLeft_Mean'] = user_df[[f'L{f}C' for f in H_freqs]].mean(axis=1)
user_df['LBRight_Mean'] = user_df[[f'R{f}C' for f in L_freqs]].mean(axis=1)
user_df['LBLeft_Mean'] = user_df[[f'L{f}C' for f in L_freqs]].mean(axis=1)

user_df['HBRight_Mean'] = user_df[[f'BR{f}C' for f in H_freqs]].mean(axis=1)
user_df['HBLeft_Mean'] = user_df[[f'BL{f}C' for f in H_freqs]].mean(axis=1)
user_df['LBRight_Mean'] = user_df[[f'BR{f}C' for f in L_freqs]].mean(axis=1)
user_df['LBLeft_Mean'] = user_df[[f'BL{f}C' for f in L_freqs]].mean(axis=1)

user_df['Ear_Asymmetry'] = np.abs(user_df['Right_Mean'] - user_df['Left_Mean'])

# Dummy placeholders for dynamic delta variables if not provided directly in single-patient form
user_df['RAC_BC'] = 0.0
user_df['LAC_BC'] = 0.0

# Align columns to match model expectations
X_user = user_df[feature_cols]

# Main Dashboard layout
col1, col2 = st.columns([1, 1])

with col1:
    st.subheader("🔮 Model Prediction")
    if st.button("Run Prediction & Compare", type="primary"):
        prediction_encoded = model.predict(X_user)[0]
        prediction_proba = model.predict_proba(X_user)[0]
        predicted_label = label_encoder.inverse_transform([prediction_encoded])[0]
        
        st.success(f"**Recommended Treatment Category:** {predicted_label}")
        
        st.write("### Class Probability Breakdown:")
        prob_df = pd.DataFrame({
            'Treatment Category': label_encoder.classes_,
            'Probability': prediction_proba
        }).sort_values(by='Probability', ascending=False)
        st.dataframe(prob_df, hide_index=True)

with col2:
    st.subheader("📊 Comparison with Historical Data")
    
    # Audiogram profile comparison plot
    fig, ax = plt.subplots(figsize=(8, 4))
    
    # Historical Mean Audiogram
    hist_right_means = [ref_df[f'R{f}C'].mean() for f in frequencies]
    hist_left_means = [ref_df[f'L{f}C'].mean() for f in frequencies]
    
    # User Audiogram
    user_right_vals = [user_inputs[f'R{f}C'] for f in frequencies]
    user_left_vals = [user_inputs[f'L{f}C'] for f in frequencies]
    
    ax.plot(frequencies, user_right_vals, marker='o', color='red', label='Patient Right Ear')
    ax.plot(frequencies, user_left_vals, marker='x', color='blue', label='Patient Left Ear')
    ax.plot(frequencies, hist_right_means, linestyle='--', color='salmon', label='Historical Avg Right')
    ax.plot(frequencies, hist_left_means, linestyle='--', color='skyblue', label='Historical Avg Left')
    
    ax.invert_yaxis()  # Audiograms standardly invert dB HL (higher down)
    ax.set_xlabel("Frequency (Hz)")
    ax.set_ylabel("Hearing Threshold (dB HL)")
    ax.set_title("Audiogram Comparison vs. Historical Dataset")
    ax.legend(fontsize=8)
    ax.grid(True)
    st.pyplot(fig)

# Distribution Comparison
st.subheader("📈 Patient Mean Hearing Threshold vs Historical Population")
fig2, ax2 = plt.subplots(figsize=(10, 3))
sns.histplot(ref_df['Right_Mean'], color='gray', label='Historical Right Ear Means', kde=True, ax=ax2, stat="density")
ax2.axvline(user_df['Right_Mean'].values[0], color='red', linestyle='-', linewidth=3, label='Current Patient Right Mean')
ax2.set_title("Patient Position within Historical Right Ear Mean Distribution")
ax2.legend()
st.pyplot(fig2)
```
